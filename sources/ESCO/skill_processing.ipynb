```python
import sys; sys.path.append("../..")

import numpy as np
import pandas as pd

from kb_graph import taxonomy_levels
import zipfile

ESCO_ZIP = "ESCO dataset - v1.2.1 - classification - en - csv.zip"


def esco_csv(name, **kwargs):
    """Read one CSV of the ESCO distribution straight out of the zip."""
    with zipfile.ZipFile(ESCO_ZIP) as archive:
        return pd.read_csv(archive.open("ESCO dataset - v1.2.1 - classification - en - csv/" + name), **kwargs)
```

```python
skills = esco_csv("skills_en.csv")
skillGroups = esco_csv("skillGroups_en.csv")
```

`skills_en.csv` holds the skills and `skillGroups_en.csv` the skill groups. Only the relevant columns are kept.

```python
skills = skills[["conceptUri", "skillType", "preferredLabel", "reuseLevel", "altLabels", "description"]]
skillGroups = skillGroups[["conceptUri", "preferredLabel", "altLabels", "description"]]
```

Strip the common URI prefix from `conceptUri`.

```python
skills["conceptUri"]= skills["conceptUri"].apply(
    lambda x: x.replace("http://data.europa.eu/esco/skill/", "")
)
skills = skills.drop_duplicates("conceptUri")

skillGroups["conceptUri"]= skillGroups["conceptUri"].apply(
    lambda x: (x
               .replace("http://data.europa.eu/esco/skill/", "")
               .replace("http://data.europa.eu/esco/isced-f/", "")
    )
)
skillGroups = skillGroups.drop_duplicates("conceptUri")
```

Split `altLabels` into a list and give each alternative label an integer id.

```python
skills["altLabels"] = skills["altLabels"].apply(
    lambda x: x.split("\n") if isinstance(x,str) else []
)
skillGroups["altLabels"] = skillGroups["altLabels"].apply(
    lambda x: x.split("\n") if isinstance(x,str) else []
)
```

```python
alt_label2id = (
    pd.concat([skills["altLabels"].explode(),skillGroups["altLabels"].explode()])
    .drop_duplicates()
    .reset_index(drop=True)
    .reset_index()
    .set_index("altLabels")
    .to_dict()["index"]
)
```

`altLabels` holds newline-separated alternative labels: split them into a list and give each label an integer id.

```python
skills["altLabels"] = skills["altLabels"].apply(lambda x: [alt_label2id[i] for i in x])
skillGroups["altLabels"] = skillGroups["altLabels"].apply(lambda x: [alt_label2id[i] for i in x])
```

Fill the missing `skillType` values with "skill/competence".

```python
nan_index = skills[skills.skillType.apply(lambda x: np.isnan(x) if isinstance(x,float) else False)].index
skills.loc[nan_index, "skillType"] = "skill/competence"
```

Rename the columns.

```python
skills = skills.rename(
    columns={
        "conceptUri": "skill_id",
        "preferredLabel": "label",
        "skillType": "skill_type",
    }
).set_index("skill_id")

skillGroups = skillGroups.rename(
    columns={
        "conceptUri": "skill_id",
        "preferredLabel": "label",
    }
).set_index("skill_id")
```

Add a column that marks leaf nodes (skills) and group nodes (skill groups).

```python
skillGroups = skillGroups.assign(skill_type="broad_skill",reuseLevel="group")
skills = pd.concat([skills, skillGroups], axis=0)
```

`languageSkillsCollection_en.csv`: the language skills. Only the relevant column is kept.

```python
lang_skill_uri = esco_csv(
    "languageSkillsCollection_en.csv",
    usecols=["conceptUri"],
    converters={"conceptUri": lambda x: x.replace("http://data.europa.eu/esco/skill/", "")}
)["conceptUri"].tolist()
```

```python
skills.loc[skills.index.isin(lang_skill_uri), "skill_type"] = "language"
```

Save the skills and the alternative labels.

```python
(pd
 .DataFrame
 .from_dict(alt_label2id, orient="index")
 .reset_index()
 .rename(columns={"index": "alt_label", 0: "alt_label_id"})
 .to_csv("processed/skills_alt_labels.csv", index=False)
 )
```

```python
# Shortest distance from a taxonomy root, plus "has no narrower concept". The skill
# pillar is a DAG, so `level` is a rank and not a tree depth (see taxonomy_levels).
level, leaf = taxonomy_levels(pd.read_csv("processed/skill_tax.csv"))
```

```python
# Indexing, not .map: a skill outside the taxonomy must raise here.
skills["level"] = [level[skill_id] for skill_id in skills.index]
skills["leaf"] = [leaf[skill_id] for skill_id in skills.index]
```
