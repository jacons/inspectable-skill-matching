```python
import sys; sys.path.append("../..")

import pandas as pd

from kb_graph import taxonomy_levels
import zipfile

ESCO_ZIP = "ESCO dataset - v1.2.1 - classification - en - csv.zip"


def esco_csv(name, **kwargs):
    """Read one CSV of the ESCO distribution straight out of the zip."""
    with zipfile.ZipFile(ESCO_ZIP) as archive:
        return pd.read_csv(archive.open("ESCO dataset - v1.2.1 - classification - en - csv/" + name), **kwargs)
```

```python
jobs = esco_csv("occupations_en.csv")
isco = esco_csv("ISCOGroups_en.csv")
```

```python
jobs = jobs[["conceptUri", "preferredLabel", "altLabels", "code", "description"]]
isco = isco[["conceptUri", "preferredLabel", "altLabels", "description"]]
```

Strip the common URI prefix from `conceptUri`.

```python
jobs["conceptUri"] = jobs["conceptUri"].apply(
    lambda x: x.replace("http://data.europa.eu/esco/occupation/", "")
)
isco["conceptUri"] = isco["conceptUri"].apply(
    lambda x: x.replace("http://data.europa.eu/esco/isco/", "")
)
jobs = jobs.drop_duplicates("conceptUri")
isco = isco.drop_duplicates("conceptUri")
```

`altLabels` holds newline-separated alternative labels: split them into a list and give each label an integer id.

```python
jobs["altLabels"] = jobs["altLabels"].apply(
    lambda x: x.split("\n") if isinstance(x, str) else []
)
isco["altLabels"] = isco["altLabels"].apply(
    lambda x: x.split("\n") if isinstance(x, str) else []
)
```

```python
alt_label2id = (
    pd.concat([jobs["altLabels"].explode(), isco["altLabels"].explode()])
    .drop_duplicates()
    .reset_index(drop=True)
    .reset_index()
    .set_index("altLabels")
    .to_dict()["index"]
)
```

Replace the alternative labels in `altLabels` with their ids.

```python
jobs["altLabels"] = jobs["altLabels"].apply(lambda x: [alt_label2id[i] for i in x])
isco["altLabels"] = isco["altLabels"].apply(lambda x: [alt_label2id[i] for i in x])
```

Rename the columns.

```python
jobs = jobs.rename(
    columns={
        "conceptUri": "occ_id",
        "preferredLabel": "label",
    }
).set_index("occ_id")

isco = isco.rename(
    columns={
        "conceptUri": "occ_id",
        "preferredLabel": "label",
    }
).set_index("occ_id")

# ISCOGroups_en.csv has a `code` column, but it holds the digits without the pillar
# letter, so the armed forces (1, 11, 110) collide with Managers (1, 11, 110). The id
# derived from the URI is the only unambiguous code these groups have.
isco["code"] = isco.index
```

Save the occupations and the alternative labels.

```python
(pd
 .DataFrame
 .from_dict(alt_label2id, orient="index")
 .reset_index()
 .rename(columns={"index": "alt_label", 0: "alt_label_id"})
 .to_csv("processed/occ_alt_labels.csv", index=False)
 )
```

```python
# One digit per ISCO level: C1 -> 1, C11 -> 2, C0110 -> 4.
isco["level"] = [len(occ_id) - 1 for occ_id in isco.index]
```

```python
# Shortest distance from a taxonomy root, plus "has no narrower concept".
level, leaf = taxonomy_levels(pd.read_csv("processed/occupation_tax.csv"))
```

```python
# Indexing, not .map: an occupation outside the taxonomy raises here.
jobs["level"] = [level[occ_id] for occ_id in jobs.index]
jobs["leaf"] = [leaf[occ_id] for occ_id in jobs.index]
isco["leaf"] = [leaf[occ_id] for occ_id in isco.index]
```

```python
# The armed-forces groups (C01, C011, C0110, C02, ...) take their id from the URI; their
# `code` would collide with Managers, Professionals and Technicians.
jobs = pd.concat([jobs, isco], axis=0)[["label", "altLabels", "code", "leaf", "level", "description"]]
assert jobs.index.is_unique, f"duplicate occ_id: {sorted(set(jobs.index[jobs.index.duplicated()]))}"
```
