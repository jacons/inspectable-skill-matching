```python
import sys; sys.path.append("../..")

import pandas as pd
import zipfile

ESCO_ZIP = "ESCO dataset - v1.2.1 - classification - en - csv.zip"


def esco_csv(name, **kwargs):
    """Read one CSV of the ESCO distribution straight out of the zip."""
    with zipfile.ZipFile(ESCO_ZIP) as archive:
        return pd.read_csv(archive.open("ESCO dataset - v1.2.1 - classification - en - csv/" + name), **kwargs)
```

# Skills

`broaderRelationsSkillPillar_en.csv`: the broader relations between skills. Only the relevant columns are kept.

```python
columns_used = ["conceptUri", "broaderUri"]
skill_rel = esco_csv("broaderRelationsSkillPillar_en.csv", usecols=columns_used)
```

Strip the common URI prefix from `conceptUri` and `broaderUri`.

```python
for col in skill_rel.columns:
    skill_rel[col] = (
        skill_rel[col]
        .str.replace("http://data.europa.eu/esco/", "", regex=False)
        .str.replace("skill/", "", regex=False)
        .str.replace("isced-f/", "", regex=False)
    )
```

Save the skill relations.

```python
skill_rel.to_csv("processed/skill_tax.csv", index=False)
```

# Skill - Occupation

`occupationSkillRelations_en.csv`: the occupation-skill relations. Only the relevant columns are kept.

```python
columns_used = ["occupationUri", "relationType", "skillUri"]

occskill_rel = esco_csv("occupationSkillRelations_en.csv", usecols=columns_used)
```

Strip the common URI prefix from `occupationUri` and `skillUri`.

```python
occskill_rel.occupationUri = occskill_rel.occupationUri.apply(
    lambda x: x.replace("http://data.europa.eu/esco/occupation/", "")
)
occskill_rel.skillUri = occskill_rel.skillUri.apply(
    lambda x: x.replace("http://data.europa.eu/esco/skill/", "")
)
```

Save the occupation-skill relations.

```python
occskill_rel.to_csv("processed/occ_skill_rel.csv", index=False)
```
