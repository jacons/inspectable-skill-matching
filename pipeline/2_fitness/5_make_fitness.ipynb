```python
import sys;sys.path.append('../..')
import pandas as pd
from pathlib import Path

from feature_engineering.fitness_matrix import build_fitness_matrix

DATA_DIR = Path("../../datasets/findhr")
```

```python
jobs = pd.read_json(DATA_DIR / "source/filtered_jobs_dt.json")[
            ["id_j", "edu_eqf", "years_experience", "languages"]
        ].set_index("id_j").add_suffix("_j").reset_index()

cvs = (pd.read_json(DATA_DIR / "source/filtered_cvs_dt.json")[
                 ["id_c", "edu_eqf", "years_experience", "languages"]
             ].set_index("id_c").add_suffix("_c").reset_index())

# The pools of 1_application.ipynb flattened to (id_j, id_c) pairs.
pairs = (pd.read_json(DATA_DIR / "applications/applications.json")[["id_j", "pool"]]
         .explode("pool").rename(columns={"pool": "id_c"})
         .astype({"id_c": int}).reset_index(drop=True))
print(len(pairs), 'pairs')

skill = pd.read_csv(DATA_DIR / 'skill_fitness.csv')
print(len(skill), 'skills')
```

```python
fv = build_fitness_matrix(pairs, jobs, cvs, skill)
fv.head()
```

```python
fv.to_csv(DATA_DIR / 'fitness_matrix.csv', index=False)
```
